# Exploratory Analysis and Visualization
 - Explored relationships between Columns.
 - Noted few intresting insights fro exploratory analysis.
 - Used various methods to capture insights from multiple columns together.

```python
# ----------------------------------------------------- UNI-VARIATE ANALYSIS ----------------------------------------------------- #
```

```python
# 1] COUNT of CONTENT TYPE.
a=sns.countplot(x='type', data=nf,width=0.3,color='skyblue',)
sns.set(style="darkgrid")
plt.title('Content Type Distribution (Movies vs TV Shows)')
plt.show()
```

```python
# Types of Shows on Netflix (pie chart)
from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot
init_notebook_mode(connected=True)
cf.go_offline()
df=pd.DataFrame(nf['type'].value_counts())
plotly.express.pie(df,names=df.index,values='count',title='Types of Shows on NETFLIX')
```

```python
# 2] TOP 10 COUNTRIES with MOST CONTENT.
top_count=nf['country'].value_counts().head(10)
sns.set(style="darkgrid")
top_count.plot(kind='barh',color='lightgreen')
plt.xlabel("COUNT")
plt.grid(axis='y')
```

```python
# 3] TOP 10 Years with MOST RELEASES.
df=pd.DataFrame(nf['release_year'].value_counts().head(10))
sns.barplot(x=df.index,y=df['count'],data=df,color='skyblue')
```

```python
# 4] Count of shows Released in years.
plt.title('Netflix Release Years')
nf["release_year"].value_counts().plot.barh(figsize=(30,20), color="#32a883")
plt.show()
```

```python
# 5] Rating Distributions
plt.figure(figsize = (10,8))
plt.title('Netflix Ratings Distribution')
sns.countplot(x='rating', data = nf)
plt.show()
```

```python
# ----------------------------------------------------- BI-VARIATE ANALYSIS ----------------------------------------------------- #
```

```python
# 1] TOP 10 Countries comparison between 'Movie' and 'Tv-Show'.
top_countries = nf['country'].value_counts().head(10).index
df_top_countries = nf[nf['country'].isin(top_countries)]
country_type_counts = pd.crosstab(df_top_countries['country'], df_top_countries['type'])
country_type_counts.plot(kind='bar', figsize=(8,6), color=['coral', 'skyblue'])
plt.title('Movies vs TV Shows in Top 10 Countries')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Type')
plt.tight_layout()
plt.show()
```

```python
# 2] Duration Analysis [MOVIES only]
movie_df = nf[nf['type'] == 'Movie']
movie_df['duration_minutes'] = movie_df['duration'].str.replace(' min','').astype(float)

plt.figure(figsize=(8,5))
sns.histplot(movie_df['duration_minutes'].dropna(), bins=30, kde=True, color='purple')
plt.title('Movie Duration Distribution')
plt.xlabel('Duration (minutes)')
plt.show()
```

```python
# 3] GENRE Word_Cloud.
from wordcloud import WordCloud
plt.figure(figsize=(10,6))
text = " ".join(nf['genre'].dropna().values)
wordcloud = WordCloud(width=800, height=400, background_color='black', colormap='Set2').generate(text)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Most Common Genres on Netflix')
plt.show()
```

```python
# 4] Movies VS TV-Shows on NETFLIX
plt.figure(figsize=(10,6))
plt.plot(nf[nf['type']=='TV Show'].groupby('release_year')['type'].count(), color='salmon', linewidth=4)
plt.plot(nf[nf['type']=='Movie'].groupby('release_year')['type'].count(), color='lightblue',linewidth=4)
plt.xlabel("Year")
plt.ylabel("Titles Added")
plt.title("Movies vs. TV Shows on Netflix\nCount of Titles Released per Year")
plt.legend(['TV Shows','Movies'])
plt.show()
```
